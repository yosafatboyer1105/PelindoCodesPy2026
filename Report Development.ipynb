# Grouping each month

```python
def groups(path, period):
    df = pd.read_excel(path, sheet_name='Accounting Month Data')
    
    old = list(df.columns)
    new = df.iloc[0]
    for i,j in zip(old,new):
        df.rename(columns={i:j}, inplace=True)
    
    df.drop(axis=0, index=0, inplace=True)
    df['Model Code_Identification ID'] = df['Model Code_Identification ID'].apply(lambda x:x.split('_'))
    
    df['Model Code'] = df['Model Code_Identification ID'].apply(lambda x:x[0])
    df['Identification ID'] = df['Model Code_Identification ID'].apply(lambda x:x[1])
    df.drop(axis=1, columns='Model Code_Identification ID', inplace=True)
    
    df['Period'] = period
    
    return df
```

```python
sep_23 = groups(r"D:\Astragraphia\BO1\Pelindo\Zweite Entwicklung\09_2023.xlsx", 'Sep 2023')
oct_23 = groups(r"D:\Astragraphia\BO1\Pelindo\Zweite Entwicklung\10_2023.xlsx", 'Okt 2023')
nov_23 = groups(r"D:\Astragraphia\BO1\Pelindo\Zweite Entwicklung\11_2023.xlsx", 'Nov 2023')
des_23 = groups(r"D:\Astragraphia\BO1\Pelindo\Zweite Entwicklung\12_2023.xlsx", 'Des 2023')
jan_24 = groups(r"D:\Astragraphia\BO1\Pelindo\Zweite Entwicklung\01_2024.xlsx", 'Jan 2024')
feb_24 = groups(r"D:\Astragraphia\BO1\Pelindo\Zweite Entwicklung\02_2024.xlsx", 'Feb 2024')
mar_24 = groups(r"D:\Astragraphia\BO1\Pelindo\Zweite Entwicklung\03_2024.xlsx", 'Mar 2024')
apr_24 = groups(r"D:\Astragraphia\BO1\Pelindo\Zweite Entwicklung\04_2024.xlsx", 'Apr 2024')
may_24 = groups(r"D:\Astragraphia\BO1\Pelindo\Zweite Entwicklung\05_2024.xlsx", 'Mei 2024')
jun_24 = groups(r"D:\Astragraphia\BO1\Pelindo\Zweite Entwicklung\06_2024.xlsx", 'Jun 2024')
jul_24 = groups(r"D:\Astragraphia\BO1\Pelindo\Zweite Entwicklung\07_2024.xlsx", 'Jul 2024')
```
