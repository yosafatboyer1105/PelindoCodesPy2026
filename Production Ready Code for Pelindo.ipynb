```python
path = r"D:\Astragraphia\BO1\Pelindo\Zweite Entwicklung\Laporan ILCS\Pre-Summary Desember 2024 - Copy.xlsx"
class excelReader:
    def __init__(self,file):
        self.file = file
    def sheetChecker(self):
        sc = pd.ExcelFile(self.file)
        print(sc.sheet_names)
    def sheetSelector(self,sheetName):
        df = pd.read_excel(self.file, sheet_name=sheetName)
        return df

reader = excelReader(path)
reader.sheetChecker()
summary = reader.sheetSelector('Summary')
details = reader.sheetSelector('Details')
print(summary.shape)
print(details.shape)
```
